## About the Dataset

The MHEALTH (Mobile HEALTH) dataset comprises body motion and vital signs recordings for ten volunteers of diverse profile while performing several physical activities. Sensors placed on the subject's chest, right wrist, and left ankle are used to measure the motion experienced by diverse body parts, namely, acceleration, rate of turn, and magnetic field orientation. The sensor positioned on the chest also provides 2-lead ECG measurements, which can be potentially used for basic heart monitoring, checking for various arrhythmias, or looking at the effects of exercise on the ECG.

## Experimental Setup

The collected dataset comprises body motion and vital signs recordings for ten volunteers of diverse profile while performing 12 physical activities (Table 1). Shimmer2 [BUR10] wearable sensors were used for the recordings. The sensors were respectively placed on the subject's chest, right wrist, and left ankle and attached by using elastic straps.

The use of multiple sensors permits us to measure the motion experienced by diverse body parts, namely, the acceleration, the rate of turn, and the magnetic field orientation, thus better capturing the body dynamics.

The sensor positioned on the chest also provides 2-lead ECG measurements which are not used for the development of the recognition model but rather collected for future work purposes. This information can be used, for example, for basic heart monitoring, checking for various arrhythmias, or looking at the effects of exercise on the ECG.

All sensing modalities are recorded at a sampling rate of 50 Hz, which is considered sufficient for capturing human activity.

## MHEALTH Dataset Data Format

The MHEALTH dataset is organized into separate log files for each subject, named `mHealth_subject<SUBJECT_ID>.log`. These files contain sensor readings and activity labels, structured as follows:

### Columns:

| Column Number | Sensor & Measurement | Description |
|---|---|---|
| 1 | Chest Sensor | Acceleration (X-axis) |
| 2 | Chest Sensor | Acceleration (Y-axis) |
| 3 | Chest Sensor | Acceleration (Z-axis) |
| 4 | ECG | Signal (Lead 1) |
| 5 | ECG | Signal (Lead 2) |
| 6 | Left Ankle Sensor | Acceleration (X-axis) |
| 7 | Left Ankle Sensor | Acceleration (Y-axis) |
| 8 | Left Ankle Sensor | Acceleration (Z-axis) |
| 9 | Left Ankle Sensor | Gyroscope (X-axis) |
| 10 | Left Ankle Sensor | Gyroscope (Y-axis) |
| 11 | Left Ankle Sensor | Gyroscope (Z-axis) |
| 12 | Left Ankle Sensor | Magnetometer (X-axis) |
| 13 | Left Ankle Sensor | Magnetometer (Y-axis) |
| 14 | Left Ankle Sensor | Magnetometer (Z-axis) |
| 15 | Right Lower Arm Sensor | Acceleration (X-axis) |
| 16 | Right Lower Arm Sensor | Acceleration (Y-axis) |
| 17 | Right Lower Arm Sensor | Acceleration (Z-axis) |
| 18 | Right Lower Arm Sensor | Gyroscope (X-axis) |
| 19 | Right Lower Arm Sensor | Gyroscope (Y-axis) |
| 20 | Right Lower Arm Sensor | Gyroscope (Z-axis) |
| 21 | Right Lower Arm Sensor | Magnetometer (X-axis) |
| 22 | Right Lower Arm Sensor | Magnetometer (Y-axis) |
| 23 | Right Lower Arm Sensor | Magnetometer (Z-axis) |
| 24 | Activity Label | (0 = Null class, other numbers represent specific activities) |


### Rows:

Each row in the file represents a single sample of data collected from all sensors at a specific time point.


### Activity Labels:

Numeric labels are used to identify the activity being performed by the subject during that sample. For example, the label '4' corresponds to the 'Walking' activity. A full list of activities and their corresponding labels is provided in the dataset documentation.

Converting all log files into csv files


```python
import pandas as pd

# Input and output file paths
input_file = 'logs_data/mHealth_subject10.log'  
output_file = 'logs_data/subject10_ecg_data.csv'

# Define column names as per documentation
columns = [
    "chest_acc_x", "chest_acc_y", "chest_acc_z",
    "ecg_1", "ecg_2",
    "la_acc_x", "la_acc_y", "la_acc_z",
    "la_gyro_x", "la_gyro_y", "la_gyro_z",
    "la_mag_x", "la_mag_y", "la_mag_z",
    "ra_acc_x", "ra_acc_y", "ra_acc_z",
    "ra_gyro_x", "ra_gyro_y", "ra_gyro_z",
    "ra_mag_x", "ra_mag_y", "ra_mag_z",
    "activity"
]

# Read the tab-separated log file
df = pd.read_csv(input_file, sep='\t', header=None, names=columns)

# Save to CSV
df.to_csv(output_file, index=False)

print(f"Converted: {output_file}")
```

```python
import pandas as pd
import glob

# Pattern to match all subject CSVs
csv_files = glob.glob("logs_data/subject*_ecg_data.csv")

# List to hold each subject's data
df_list = []

for file in csv_files:
    df = pd.read_csv(file, header=None)
    df_list.append(df)

# Combine all into one DataFrame
full_df = pd.concat(df_list, ignore_index=True)

# Optional: assign column names if you haven't yet
full_df.columns = [
    "acc_chest_x", "acc_chest_y", "acc_chest_z",
    "ecg_1", "ecg_2",
    "acc_ankle_x", "acc_ankle_y", "acc_ankle_z",
    "gyro_ankle_x", "gyro_ankle_y", "gyro_ankle_z",
    "magnet_ankle_x", "magnet_ankle_y", "magnet_ankle_z",
    "acc_arm_x", "acc_arm_y", "acc_arm_z",
    "gyro_arm_x", "gyro_arm_y", "gyro_arm_z",
    "magnet_arm_x", "magnet_arm_y", "magnet_arm_z",
    "label"
]
```

```python
full_df.to_csv("combined_mhealth_data.csv", index=False)
```
