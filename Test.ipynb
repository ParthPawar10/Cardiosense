```python
import pandas as pd
from datetime import datetime
```

```python
df_original = pd.read_csv('combined_mhealth_data.csv')
```

```python
def save_test_case_by_label(label_value):
    if label_value not in range(1, 13):
        raise ValueError("label_value must be between 1 and 12")

    filtered_df = df_original[df_original['label'] == label_value]

    if filtered_df.empty:
        print(f"No rows found with label {label_value}")
        return

    test_case = filtered_df.sample(n=1, random_state=42)
    filename = f"ecg_{datetime.today().strftime('%Y-%m-%d')}.csv"
    test_case.to_csv(filename, index=False)
    print(f"✅ Test case with label {label_value} saved to {filename}")
```

```python
save_test_case_by_label(2)
```
